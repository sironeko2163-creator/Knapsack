```python
import random

# 品物ごとの重さと価値
weights = [2, 3, 4, 5, 9]
values  = [3, 4, 5, 8, 10]

capacity = 10          # 重量上限
population_size = 10   # 個体数
generations = 30       # 世代数
mutation_rate = 0.05   # 突然変異率


# 個体の適応度を計算
def fitness(individual):
    total_weight = sum(
        individual[i] * weights[i]
        for i in range(len(weights))
    )

    total_value = sum(
        individual[i] * values[i]
        for i in range(len(values))
    )

    # 重量上限を超えた個体は評価0
    if total_weight > capacity:
        return 0

    return total_value


# 初期個体群をランダムに作る
population = [
    [random.randint(0, 1) for _ in weights]
    for _ in range(population_size)
]


for generation in range(generations):

    # 適応度が高い順に並べる
    population.sort(
        key=fitness,
        reverse=True
    )

    # 上位半分を親候補にする
    parents = population[:population_size // 2]

    # 最優秀個体をそのまま次世代へ残す
    next_population = [population[0][:]]

    while len(next_population) < population_size:

        # 親を2個体選ぶ
        parent1, parent2 = random.sample(parents, 2)

        # 一点交叉
        point = random.randint(1, len(weights) - 1)
        child = parent1[:point] + parent2[point:]

        # 突然変異
        for i in range(len(child)):
            if random.random() < mutation_rate:
                child[i] = 1 - child[i]

        next_population.append(child)

    population = next_population


# 最終結果
best = max(population, key=fitness)

total_weight = sum(
    best[i] * weights[i]
    for i in range(len(weights))
)

print("最良個体:", best)
print("総重量:", total_weight)
print("総価値:", fitness(best))
```
